```python
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#turning into col names in small letters and using underscore instead of space
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age group
labels = ['Young Adults', 'Adult', 'Middle-aged', 'Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']]
```

```python
#purchase frequency days-create column
frequency_mapping = {'Fortnightly':14,
                     'Weekly':7,
                     'Monthly':30,
                     'Quarterly':90,
                     'Bi-Weekly':14,
                     'Annually':365,
                     'Every 3 Months':90}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine
import urllib

# SQL Server details
server = r'LAPTOP-IKD10G32\SQLEXPRESS'     # Replace with your server
database = 'customer_behaviour'            # Replace with your database name

# Windows Authentication
params = urllib.parse.quote_plus(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

print("Connection successful!")
```

```python
from sqlalchemy import text

with engine.connect() as conn:
    result = conn.execute(text("SELECT DB_NAME()"))
    print(result.fetchone())
```

```python
df.to_sql(
    name="customer",
    con=engine,
    if_exists="replace",
    index=False
)

print("Table created successfully!")
```
